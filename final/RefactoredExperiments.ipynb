```python
from refactored.pathfinder import * 
from refactored.graphs import * 
from refactored.algorithms import * 

test_graph = HeuristicGraph()

# Add nodes
for i in [1, 2, 3, 4]:
    test_graph.add_node(i)

# Add edges
test_graph.add_edge(1, 2, 1.0)
test_graph.add_edge(2, 3, 2.0)
test_graph.add_edge(3, 4, 1.0)
test_graph.add_edge(1, 4, 5.0) # The "deceptive" direct route

# Setup the A* Heuristic (Estimates to Destination Node 4)
# For A* to work, the heuristic must be <= the true distance
dummy_heuristic = {
    1: 3.0, # Guessing it takes 3 to get to 4
    2: 2.0, # Guessing it takes 2 to get to 4
    3: 1.0, # Guessing it takes 1 to get to 4
    4: 0.0  # Already there
}
test_graph.set_heuristic(dummy_heuristic)

# ==========================================
# 2. Setup the Context (Pathfinder)
# ==========================================
# We instantiate the Context object and give it our graph
finder = ShortPathFinder()
finder.set_graph(test_graph)

# Define our test variables
source_node = 1
dest_node = 4
expected_distance = 4.0

print("--- Starting Architecture Test ---\n")

# ==========================================
# 3. Test Dijkstra (The Baseline)
# ==========================================
finder.set_algorithm(Dijkstra())
dijkstra_result = finder.calc_short_path(source_node, dest_node)

print(f"Dijkstra Result: {dijkstra_result}")
assert dijkstra_result == expected_distance, "Dijkstra failed!"

# ==========================================
# 4. Test A* (The Adapter)
# ==========================================
finder.set_algorithm(A_Star())
astar_result = finder.calc_short_path(source_node, dest_node)

print(f"A* Result:       {astar_result}")
assert astar_result == expected_distance, "A* failed!"

# ==========================================
# 5. Test Bellman-Ford (The Brute Force)
# ==========================================
finder.set_algorithm(Bellman_Ford())
bf_result = finder.calc_short_path(source_node, dest_node)

print(f"Bellman-Ford:    {bf_result}")
assert bf_result == expected_distance, "Bellman-Ford failed!"

# ==========================================
# 6. Final Verdict
# ==========================================
print("\nSUCCESS: All algorithms implemented the SPAlgorithm interface correctly")
print("SUCCESS: The ShortPathFinder successfully swapped strategies at runtime")
print("SUCCESS: A_Star successfully extracted the heuristic from HeuristicGraph")
```
