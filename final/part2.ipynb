**Discussion Questions**

*What issues with Dijkstra's algorithm is A* trying to address?*

The core problem with Dijkstra's is that it uniformly explores every direction until it finds the shortest path. It has no "sense of direction" so to speak as it does not know where the goal is. A* addresses this by keeping track of the distance to the goal node and guides/directs Dijkstra's towards the goal, thus reducing inefficient node exploration.

*How would you empirically test Dijkstra vs A*?

You could empirically test Dijkstra vs A* by running both algorithms on the same sets of graphs and recording both runtime and nodes explored. Repeat this with varying graph sizes, density, and start/end goals to obtain a broader range of data.

*If you generated an arbitrary heuristic function (similar to randomly generating weights), how would Dijkstra’s algorithm compare to A*

I think if you generated an arbitrary heuristic function, Dijkstra's would perform better than A* in almost all cases. This is because the whole point of A* is that it has a good heuristic function that allows it to perform more efficiently than Dijkstra's. If you had a random heuristic function, you would have no idea how A* would behave, and more likely than not it will be relying on a poor heuristic function that actively harms the algorithm's efficiency. It would still *find* a path, but it may not be the shortest path and the algorithm may explore a massive and unnecessary amount of nodes.

*When would you use A* instead of Dijkstra?*

I would use A* when I haveaa meaningful estimate to the goal (heuristic). For example, I think A* excels in GPS navigation because there is a clear, measurable distance to the goal which can help guide the algorithm to the most optimal path and also avoiding inefficient path exploration. Also, if you had a reliable heuristic and time is of the essence then A* is clearly the better option. For example, in robotics, a robot needs to react instantly to environment changes and path planning should happen as fast as possible. Using spatial heuristics, A* can gets a path fast enough for real time movement.
