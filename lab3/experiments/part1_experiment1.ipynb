```python
import random
import numpy as np
from auxiliary.lab3 import *
from auxiliary.bst import *

n = 10000   # list length
trials = 100      # number of random lists

rbt_heights = []
bst_heights = []
differences = []

for t in range(trials):
    nums = random.sample(range(1, 10 * n), n)   # unique random integers

    rbt = RBTree()
    bst = BST()
    for i in nums:
        rbt.insert(i)
        bst.insert(i)

    rh = rbt.get_height()
    bh = bst.get_height()
    rbt_heights.append(rh)
    bst_heights.append(bh)
    differences.append(bh - rh)


print()
print(f"  Lists: {trials} random lists of {n:,} integers")
print(f"  Avg RBT height  : {np.mean(rbt_heights):.2f}")
print(f"  Avg BST height  : {np.mean(bst_heights):.2f}")
print(f"  Avg difference  : {np.mean(differences):.2f}  (BST - RBT)")
```

**Brief Discussion**


**Is there a case you would prefer a BST over an RBT for trees of similar height?**

I think the reason we may prefer a BST over an RBT is intuitively similar as to why we would prefer insertion sort over merge sort/quick sort. I think the core idea comes down to the question; what situations will the overhead (additional memory, complex implementation, potential debugging) of implementing and maintaing an RBT outweigh the benefits?

If the input is randomly ordered, a BST will naturally produce a reasonably balanced tree with expected height O(log n), meaning the self-balancing properties of an RBT provide little relative benefit. In this case, the simpler BST insert (a plain traversal) will be faster per operation. In addition, if the task is simpler such as making many lookups but not many insertions/deletions, a BST will perform very similarly to an RBT while also avoiding the associated overhead.
