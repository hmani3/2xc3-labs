```python
# i used different variables i used k instead of i becasue the double use of i was confusing and i wanted to make it more clear.
def child_degree(k): # what degree does the k-th child have
    if k <= 2:
        return 0
    return k - 2

def xc3_height(deg): # whats the height of the tree
    if deg <= 0:  # base case if no children
        return 1

    biggest_child_height = 0
    for k in range(1, deg + 1): # for each child in postiion k
        cd = child_degree(k)     # we compute the child's degree (cd)        
        h = xc3_height(cd)       # we compute the height of the child (h)
        if h > biggest_child_height:   # track the max child height seen so far
            biggest_child_height = h

    return 1 + biggest_child_height # the height is th 1 + the max of each chiald subtree

def xc3_nodes(deg): # number of nodes in the tree
    if deg <= 0:  # base case
        return 1

    total = 1  # root of the tree
    for k in range(1, deg + 1):
        cd = child_degree(k)  #same logic as above.
        total += xc3_nodes(cd)

    return total
# output  
for i in range(26):
    print("degree =", i, "height =", xc3_height(i), "nodes =", xc3_nodes(i))
```

Experiment 3 (Height h(i))
Now lets look at height. Height we have 1, 2, 2, 3, 3, 4, 4, 5, 5... what we can see is height increases every 2 degrees. So h(0) = 1, h(1) = 2, and h(i) = 1 + h(i-2) for i >= 2. Why though? The tallest path always goes down the right most child (based on the XC3 definition). Therefore the deepest subtree is always the right most with the largest degree i - 2. Each time we move down the right most child, the degree drops by 2, so the height goes up by 1 every 2 degrees then.

Experiment 4 (nodes(i) and O(log n))
Looking at the output it’s pretty obvious the nodes follow a Fibonacci sequence (1, 2, 3, 5, 8, 13, 21...). So we can write nodes(0) = 1, nodes(1) = 2, and nodes(i) = nodes(i-1) + nodes(i-2) for all i >= 2. Why though? When we increase the degree from i - 1 to i, we add exactly one new child subtree at the end, and that subtree has degree (i-2). So we end up with nodes(i) - nodes(i-1) = nodes(i-2). Doing a little algebra, we get nodes(i) = nodes(i-1) + nodes(i-2). So then comparing this against O(log n), we know Fibonacci grows exponentially. In particular Fibonacci numbers grow proportionally to the golden ratio (phi^i). That means the nodes(i) grows like phi^i, up to a constant factor, lets call it C. If we flip that around we get that a tree with n nodes has n >= C * phi^i, so then using log laws i <= log base phi (n/C), which is the same as i <= log base phi n up to a constant shift. Due to how bases are changed in logarithms, showing that i is O(log base phi n) means the same as this being O(log n) for any other constant base. Therefore we can conclude that since height grows with the degree, it only increases by 1 every time the degree increases by 2, the height is also O(log n).
