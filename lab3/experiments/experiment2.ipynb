```python
import random
import matplotlib.pyplot as plt
from auxiliary.lab3 import *
from auxiliary.bst import *
from auxiliary.rbt import *

n = 1000
runs = 20
swap_values = [0,1,2,5,10,25,50,100,200,500]

results = []

for swaps in swap_values:

    total_diff = 0

    for _ in range(runs):

        arr = list(range(n))

        # perform swaps
        for _ in range(swaps):
            i = random.randint(0, n-1)
            j = random.randint(0, n-1)
            arr[i], arr[j] = arr[j], arr[i]

        bst = BST()
        rbt = RBTree()

        for v in arr:
            bst.insert(v)
            rbt.insert(v)

        diff = bst.get_height() - rbt.get_height()
        total_diff += diff

    avg_diff = total_diff / runs
    results.append(avg_diff)

plt.plot(swap_values, results, marker='o')
plt.xlabel("Number of Swaps")
plt.ylabel("Average Height Difference (BST - RBT)")
plt.title("Effect of Disorder on Tree Height Difference")
plt.show()
```

**Outline**

List length: 1,000 integers

Trials per swap count: 50

Swap counts tested: 0, 1, 2, 3, 4, 5, 10, 25, 50, 100, 250, 500, 1000, 2500, 5000

Each trial starts with a perfectly sorted list, applies the given number of random swaps, then inserts into both an RBT and BST, then recording both heights*

**Brief Discussion**

Overall, the experiment demonstrates that BST performance is highly sensitive to input order, while RBT's remain balanced regardless of how sorted the input is. Increasing the number of swaps makes the BST behave more like a randomly built tree, reducing the height difference, but the RBT still provides consistently balanced performance.
