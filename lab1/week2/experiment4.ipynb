# Experiment 4 (Week 2)

**Goal:** Compare the 3 *good* sorting algorithms (Merge, Heap, Quick) on random lists.



```python
import time
import statistics
import matplotlib.pyplot as plt

from good_sorts import mergesort, heapsort, quicksort
from bad_sorts import create_random_list  
```

```python
def time_sort(sort_fn, arr, runs=5):
    times = []
    for _ in range(runs):
        data = arr[:]  
        t0 = time.perf_counter()
        sort_fn(data)
        t1 = time.perf_counter()
        times.append(t1 - t0)
    return min(times)  
```

```python
runs = 5
max_value = 10**6
n_values = [1000, 2000, 5000, 10000, 20000]
```

```python
merge_times = []
heap_times  = []
quick_times = []

for n in n_values:
    base = create_random_list(n, max_value)

    merge_times.append(time_sort(mergesort, base, runs))
    heap_times.append(time_sort(heapsort, base, runs))
    quick_times.append(time_sort(quicksort, base, runs))

merge_times, heap_times, quick_times
```

```python
plt.figure()
plt.plot(n_values, merge_times, label="Merge")
plt.plot(n_values, heap_times,  label="Heap")
plt.plot(n_values, quick_times, label="Quick")
plt.xlabel("List length (n)")
plt.ylabel("Time (seconds)")
plt.title("Experiment 4: Merge vs Heap vs Quick (random lists)")
plt.legend()
plt.show()
```

# Executive Summary:
