# Experiment 4 (Week 2)

**Goal:** Compare the 3 *good* sorting algorithms (Merge, Heap, Quick) on random lists.



```python
import time
import matplotlib.pyplot as plt
import gc

from auxilary.good_sorts import mergesort, heapsort, quicksort
from auxilary.bad_sorts import create_random_list  
```

```python
def time_sort(sort_fn, arr, runs=5):
    times = []
    for _ in range(runs):
        
        data = arr[:]  
        gc.disable()
        t0 = time.perf_counter()
        sort_fn(data)
        t1 = time.perf_counter()
        gc.enable()
        times.append(t1 - t0)
        
    return min(times)  
```

```python
runs = 5
max_value = 10**6
length = 20000
```

```python
merge_times = []
heap_times  = []
quick_times = []

for i in range(100, length + 1, 100):
    base = create_random_list(i, max_value)
    merge_times.append(time_sort(mergesort, base, runs))
    heap_times.append(time_sort(heapsort, base, runs))
    quick_times.append(time_sort(quicksort, base, runs))
```

```python

lengths = [i * 100 for i in range(len(merge_times))]
plt.figure()
plt.plot(lengths, merge_times, label="Merge")
plt.plot(lengths, heap_times,  label="Heap")
plt.plot(lengths, quick_times, label="Quick")
plt.xlabel("List length (n)")
plt.ylabel("Time (seconds)")
plt.title("Experiment 4: Merge vs Heap vs Quick (random lists)")
plt.legend()
plt.show()
```

# Executive Summary:
In experiment 4, I comapred the runtimes of merge, heap and quick sort on random lists from with size n. For each sized list i ran each algorithm to see how it performs, and reporte the min runtime. The graph shows that with n increase the runtime for all 3 algrithms increase, and the quick sort algorithm was generally the fastest reagrdless of the range of values tested, Overall the results match the expectation that these good sorting algorithms run at around O(n logn n) time on random imputs which is much better than the "bad sorts".
