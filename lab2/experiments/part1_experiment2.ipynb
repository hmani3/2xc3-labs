```python
from auxillary.graph import *
from auxillary.helper import *
import matplotlib.pyplot as plt
import numpy as np
```

```python
def run_connectivity_experiment(node_counts = [10, 50, 100], trials = 1000):
    
    
    # Sweep the ratio of (edges / nodes) from 0 to 4.0 in 40 steps
    edge_ratios = np.linspace(0, 4.0, 40) 
    
    plt.figure(figsize=(10, 6))
    
    for i in node_counts:
        probabilities = []
        print(f"Running connectivity simulation for {i} nodes...")
        
        for ratio in edge_ratios:
            j = int(i * ratio)
            connected_count = 0
            
            for _ in range(trials):
                G = create_random_graph(i, j)
                if is_connected(G):
                    connected_count += 1
            
            probabilities.append(connected_count / trials)
            
        plt.plot(edge_ratios, probabilities, marker='o', label=f'{i} nodes')

    plt.title('Probability of Graph Connectivity vs. Edge-to-Node Ratio')
    plt.xlabel('Ratio of Edges to Nodes (j / i)')
    plt.ylabel('Probability of being Connected')
    
    plt.legend()
    plt.grid(True)
    plt.show()
```

```python
run_connectivity_experiment(trials = 5000)
```
