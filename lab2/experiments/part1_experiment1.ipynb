```python
from auxillary.graph import *
from auxillary.helper import *
import matplotlib.pyplot as plt
import numpy as np
```

```python
def run_cycle_experiment(node_counts = [10, 50, 100], trials = 1000):    
    edge_ratios = np.linspace(0, 1.5, 30) 
    
    plt.figure(figsize=(10, 6))

    for i in node_counts:
        probabilities = []
        print(f"Running simulation for {i} nodes...")
        
        for ratio in edge_ratios:
            # Calculate absolute number of edges (j) based on the ratio
            j = int(i * ratio)
            cycle_count = 0
            
            for _ in range(trials):
                G = create_random_graph(i, j)
                if has_cycle(G):
                    cycle_count += 1
            
            # Calculate probability for this specific point
            prob = cycle_count / trials
            probabilities.append(prob)
            
        # lot the curve for this node count
        plt.plot(edge_ratios, probabilities, marker='o', label=f'{i} nodes')

    plt.title('Probability of a Cycle vs. Edge-to-Node Ratio')
    plt.xlabel('Ratio of Edges to Nodes (j / i)')
    plt.ylabel('Probability of a Cycle')
    
    plt.axvline(x=0.5, color='red', label='edge / node (j/i = 0.5)')
    
    plt.legend()
    plt.grid(True)
    plt.show()
```

```python
run_cycle_experiment([10, 50, 100, 1000, 5000], 1000)
```
