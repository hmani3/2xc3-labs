```python
from auxillary.graph import *
from auxillary.helper import *
import matplotlib.pyplot as plt
import timeit
import gc
```

```python
def cycle_experiment(nodes=100, runs=50):
    max_edges = nodes * (nodes - 1) // 2
    
    # Spread edge counts from very sparse → dense
    edge_values = range(0, max_edges + 1, max_edges // 20)

    probabilities = []

    for edges in edge_values:
        cycle_count = 0

        for _ in range(runs):
            G = create_random_graph(nodes, edges)

            if has_cycle(G):
                cycle_count += 1

        probabilities.append(cycle_count / runs)

    return list(edge_values), probabilities


edges, probs = cycle_experiment()

plt.plot(edges, probs)
plt.xlabel("Number of edges")
plt.ylabel("Probability of cycle")
plt.title("Cycle probability vs number of edges")
plt.show()
```

**Conclusion**

The experiment shows that the probability of a cycle increases rapidly as the number of edges increases. Sparse graphs are unlikely to contain cycles, while denser graphs practically always do. This makes sense since trees (which have no cycles) have exactly n−1 edges, and adding extra edges tends to introduce cycles.
