```python
from auxillary.graph import *
from auxillary.helper import *
import matplotlib.pyplot as plt
import timeit
import gc
```

```python
def cycle_experiment(nodes=100, runs=50):
    max_edges = nodes * (nodes - 1) // 2
    edge_values = range(0, nodes * 2, 5)  

    probabilities = []

    for edges in edge_values:
        cycle_count = 0

        for _ in range(runs):
            G = create_random_graph(nodes, edges)
            if has_cycle(G):
                cycle_count += 1

        prob = cycle_count / runs
        probabilities.append(prob)

    return edge_values, probabilities

edges, probs = cycle_experiment()

plt.plot(edges, probs)
plt.xlabel("Number of edges")
plt.ylabel("Probability of cycle")
plt.title("Cycle probability vs number of edges")
plt.show()
```

**Experiment Outline**

We conducted the following experiment:
- Number of nodes (n): 100
- Number of edges (m): varied from 0 to 195 in increments of 5
- Number of runs per edge value: 50 random graphs

For each value of m:
We generated 50 random undirected graphs with 100 nodes and m edges.

Graphs were generated without:
- Self-loops
- Duplicate edges

We used a helper function has_cycle to detect whether the graph contains a cycle.

We computed the proportion:
Cycle Probability = Number of graphs with a cycle / 50



These values were then plotted on a graph.

**Conclusion**

The experimental results show that for graphs with 100 nodes, the probability of containing a cycle increases very drastically as the number of edges increases. Graphs with no edges obviously contain no cycles, but adding even a small number of edges makes the presence of a cycle extremely likely. This seems to suggest that in large random graphs, cycles can exist quickly as edges are added. These results also conform with theoretical expectations that trees are rare among random graphs unless the number of edges is carefully restricted.
